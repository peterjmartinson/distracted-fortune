```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skewnorm, norm

print(np.__version__)
```

```python


# Set style
sns.set(style="whitegrid")

# Generate data
np.random.seed(42)

# Normal distribution
normal_data = np.random.normal(loc=0, scale=1, size=1000)

# Left-skewed distribution (using skewnorm from scipy.stats)
left_skewed_data = skewnorm.rvs(a=-100, loc=-1, scale=1, size=1000)  # 'a' controls skewness

# Plot the distributions
plt.figure(figsize=(10, 6))
sns.kdeplot(normal_data, label="Normal Distribution", shade=True)
sns.kdeplot(left_skewed_data, label="Left-Skewed Distribution", shade=True)

# Add labels and title
plt.title("Comparison of Normal and Left-Skewed Distributions", fontsize=16)
plt.xlabel("Value", fontsize=14)
plt.ylabel("Density", fontsize=14)
plt.legend()

# Show plot
plt.show()
```

```python
# Set style
sns.set(style="whitegrid")

# Generate data
np.random.seed(42)

# Normal distribution with mean=3 and std=1
normal_data = np.random.normal(loc=3, scale=1, size=10000)

# Left-skewed distribution (using skewnorm from scipy.stats)
skewness = 10  # Strong left skew
left_skewed_data = skewnorm.rvs(a=skewness, loc=0, scale=1, size=10000)

# Shift the left-skewed data to ensure the minimum value is 0
left_skewed_data = left_skewed_data - np.min(left_skewed_data)

# Plot the distributions
plt.figure(figsize=(10, 6))
sns.kdeplot(normal_data, label="Normal Distribution (Mean=3)", shade=True)
sns.kdeplot(left_skewed_data, label="Left-Skewed Distribution (Min=0)", shade=True)

# Add labels and title
plt.title("Comparison of Normal and Left-Skewed Distributions", fontsize=16)
plt.xlabel("Value", fontsize=14)
plt.ylabel("Density", fontsize=14)
plt.legend()

# Show plot
plt.show()
```

```python
# Normal distribution PDF
def normal_pdf(x, mu=3, sigma=1):
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

# Skew-normal distribution PDF (shifted)
def skew_normal_pdf(y, alpha=10, min_x=0):
    z = y + min_x
    return 2 * norm.pdf(z) * norm.cdf(alpha * z)

# Example usage
x_values = np.linspace(0, 6, 1000)
normal_values = normal_pdf(x_values)
skew_normal_values = skew_normal_pdf(x_values)

# Plot the PDFs
plt.figure(figsize=(10, 6))
plt.plot(x_values, normal_values, label="Normal Distribution (Mean=3)")
plt.plot(x_values, skew_normal_values, label="Left-Skewed Distribution (Min=0)")
plt.title("PDFs of Normal and Left-Skewed Distributions", fontsize=16)
plt.xlabel("Value", fontsize=14)
plt.ylabel("Density", fontsize=14)
plt.legend()
plt.show()
```
