```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skewnorm

# Set style
sns.set(style="whitegrid")

# Generate data
np.random.seed(42)

# Normal distribution
normal_data = np.random.normal(loc=0, scale=1, size=1000)

# Left-skewed distribution (using skewnorm from scipy.stats)
left_skewed_data = skewnorm.rvs(a=10, loc=0, scale=1, size=1000)  # 'a' controls skewness

# Plot the distributions
plt.figure(figsize=(10, 6))
sns.kdeplot(normal_data, label="Normal Distribution", shade=True)
sns.kdeplot(left_skewed_data, label="Left-Skewed Distribution", shade=True)

# Add labels and title
plt.title("Comparison of Normal and Left-Skewed Distributions", fontsize=16)
plt.xlabel("Value", fontsize=14)
plt.ylabel("Density", fontsize=14)
plt.legend()

# Show plot
plt.show()
```
